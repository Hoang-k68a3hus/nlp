```python
import re

text = "The cats are playing."
words = re.findall(r"[A-Za-z]+", text)

try:
    from transformers import AutoTokenizer
    tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
    encoded = tokenizer(text, add_special_tokens=True)
    ids = encoded["input_ids"]
    tokens = tokenizer.convert_ids_to_tokens(ids)
    mask = encoded["attention_mask"]
    pieces = [tokenizer.tokenize(word) for word in words]
    method = "bert-base-uncased (AutoTokenizer)"
except (ImportError, OSError):
    vocab = {"[CLS]": 101, "the": 1996, "cats": 8870, "are": 2024, "playing": 2652, ".": 1012, "[SEP]": 102}
    tokens = ["[CLS]", *re.findall(r"[a-z]+|[^\w\s]", text.lower()), "[SEP]"]
    ids = [vocab[token] for token in tokens]
    mask = [1] * len(tokens)
    pieces = [[word.lower()] for word in words]
    method = "Bảng tra BERT offline cho câu mẫu (không phải AutoTokenizer)"

special = [(token, ids[i]) for i, token in enumerate(tokens) if token in ("[CLS]", "[SEP]")]
print("Tokenizer:", method)
print("Text:", text)
print("Token sequence:", tokens)
print("Token IDs:", ids)
print("Special tokens:", special)
print("Attention mask:", mask)
print("Words:", len(words))
print("Tokens (không tính special):", len(tokens) - len(special))
print("Tokens (có special):", len(tokens))
print("WordPiece pieces của các từ:", sum(map(len, pieces)))
print("Subword tách thêm:", sum(len(p) - 1 for p in pieces))
```
